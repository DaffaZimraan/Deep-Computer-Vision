```python
# Import Library
import cv2
from matplotlib import pyplot as plt
```

```python
# Read and Display Image
img = cv2.imread('assets/image.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
# Convert to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis('off')
plt.show()
```

```python
# Apply Gaussian Blur to the image
blur = cv2.GaussianBlur(img, (7, 7), 0)  # (7, 7) is the kernel size, 0 means auto standard deviation
plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.title("Blur")
plt.axis('off')
plt.show()
```

```python
# Apply Canny Edge Detection to the image
edges = cv2.Canny(img, 100, 200)  # 100 is the lower threshold, 200 is the upper threshold for edge detection
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.axis('off')
plt.show()
```
